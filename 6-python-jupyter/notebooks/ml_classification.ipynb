# Handwritten digit classification

Trains a small classifier on the `digits` dataset (8x8 greyscale images of handwritten digits 0-9) using **scikit-learn**.

If you get a `ModuleNotFoundError: No module named 'sklearn'`, your container is missing `scikit-learn` from `requirements.txt` — that's exactly the kind of version pain containers fix.

```python
import numpy as np
import matplotlib.pyplot as plt
import sklearn
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay

print('numpy       ', np.__version__)
print('scikit-learn', sklearn.__version__)
```

```python
digits = load_digits()
print('shape:', digits.images.shape)
print('classes:', digits.target_names)

fig, axes = plt.subplots(1, 8, figsize=(10, 2))
for ax, img, label in zip(axes, digits.images, digits.target):
    ax.imshow(img, cmap='gray_r')
    ax.set_title(label)
    ax.axis('off')
plt.tight_layout()
plt.show()
```

```python
X = digits.images.reshape(len(digits.images), -1)
y = digits.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

model = RandomForestClassifier(n_estimators=200, random_state=42)
model.fit(X_train, y_train)
```

```python
preds = model.predict(X_test)
print(f'Accuracy: {accuracy_score(y_test, preds):.4f}')

cm = confusion_matrix(y_test, preds)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=digits.target_names)
disp.plot(cmap='Blues', colorbar=False)
plt.title('Confusion matrix')
plt.tight_layout()
plt.show()
```

```python
i = 7
sample = X_test[i].reshape(8, 8)
predicted = model.predict([X_test[i]])[0]
actual = y_test[i]

plt.imshow(sample, cmap='gray_r')
plt.title(f'predicted: {predicted}   actual: {actual}')
plt.axis('off')
plt.show()
```
