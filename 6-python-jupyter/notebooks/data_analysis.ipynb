# Tiny sales analysis

Reads `/data/sales.csv` and produces some basic stats + a plot.

If the imports below fail with a `ModuleNotFoundError`, your container is missing dependencies
from `requirements.txt`.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print('pandas', pd.__version__)
print('numpy ', np.__version__)
```

```python
df = pd.read_csv('/data/sales.csv', parse_dates=['date'])
df['revenue'] = df['units'] * df['unit_price']
df.head()
```

```python
by_region = df.groupby('region')['revenue'].sum().sort_values(ascending=False)
by_region
```

```python
ax = by_region.plot(kind='bar', title='Revenue by region')
ax.set_ylabel('Revenue')
plt.tight_layout()
plt.show()
```
